```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

np.random.seed(42)
n_racers = 5

speeds_a = np.random.normal(loc=8.5, scale=0.6, size=n_racers)
speeds_b = np.random.normal(loc=7.2, scale=0.8, size=n_racers)

t_stat, p_value = stats.ttest_ind(speeds_a, speeds_b)

fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

frames = 40
max_distance = 100

def update(frame):
    ax.clear()
    t = frame / 4.0

    pos_x_a = np.minimum(speeds_a * t, max_distance)
    pos_y_a = np.ones(n_racers) * 1
    pos_z_a = np.sin(pos_x_a * 0.3) + 1

    pos_x_b = np.minimum(speeds_b * t, max_distance)
    pos_y_b = np.ones(n_racers) * 2
    pos_z_b = np.cos(pos_x_b * 0.3) + 1

    ax.scatter(pos_x_a, pos_y_a, pos_z_a, color='dodgerblue', s=100, label='Team A (Racers)')
    ax.scatter(pos_x_b, pos_y_b, pos_z_b, color='crimson', s=100, label='Team B (Racers)')

    ax.set_xlim(0, max_distance + 10)
    ax.set_ylim(0, 3)
    ax.set_zlim(0, 3)

    ax.set_xlabel('Race Track (Distance)')
    ax.set_ylabel('Team Lanes')
    ax.set_zlabel('Height Variation')

    status = "Significant!" if p_value < 0.05 else "Not Significant"
    ax.set_title(f"3D Object Race & T-Test Comparison\np-value: {p_value:.4f} ({status})", fontsize=11)
    ax.legend(loc='upper left')

ani = FuncAnimation(fig, update, frames=frames, interval=120)
plt.close()

HTML(ani.to_html5_video())
```
